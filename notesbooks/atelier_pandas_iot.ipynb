5) Créer, de deux façon différentes, une Series de 4 températures avec comme index les heures de
mesure 12h, 13h, 14h et 15h

```python
#premiere facon
temperature = pd.Series([34.5, 23.4, 19.8, 34.2], 
index=["12h", "13h", "14h", "15h"])

print(temperature)
```

```python
#deuxieme facon
temperature = pd.Series({
  "12h" : "34.5",
  "13h" : "23.4",
  "14h" : "19.8",
  "15h" : "34.2",
 })

print(temperature)
 
```

2) Ajouter la colonne "niveau_temperature" pour catégoriser la température en
"Élevée" (température > 30) ou "Normale". On peut utiliser la méthode where() de numpy
